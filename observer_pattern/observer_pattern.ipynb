#### so basically change in state of an object is notified to other objects
##### this notification is triggered by the object itself whose state is notified to others, its called subject in this pattern

```python
from abc import ABC, abstractmethod
class central_government():
    def __init__(self):
        self.gst_tax = "18%"
        self.state_governments = []
    
    def add_state_govt(self, state_govt):
        self.state_governments.append(state_govt)

    def update_gst_tax(self, new_rate):
        self.gst_tax = new_rate
        self.notify_state_govts_on_gst_change()

    def notify_state_govts_on_gst_change(self):
        for each_state_govt in self.state_governments:
            each_state_govt.update_central_govt_gst_rate(self.gst_tax)

class state_government():
    def __init__(self, name):
        self.state_name = name 
        self.central_govt_gst_rate = "18%"

    def update_central_govt_gst_rate(self, central_govt_gst_rate):
        self.central_govt_gst_rate = central_govt_gst_rate
        print(f"{self.state_name} succesfully updated its gst rate to {self.central_govt_gst_rate}")

india = central_government()

karnataka = state_government("karnataka")
kerala = state_government("kerala")

india.add_state_govt(karnataka)
india.add_state_govt(kerala)

india.update_gst_tax("20%")
```

### basically any change in the central_govt is notified to the state governments, obviously state governments should be added to central govt's list
