```python
class animal():
    def __init__(self, **kwargs):
        self.weight = kwargs.get("weight")
        self.skin_color = kwargs.get("skin_color")

cat = animal(weight=25, skin_color = "white")
vars(cat)
```

```python
from abc import  ABC, abstractmethod

class animal(ABC):
    def __init__(self, weight, height):
        self.height=height
        self.weight=weight

    @abstractmethod
    def make_sound(self):
        pass

class elephant(animal):
    def __init__(self, weight, height, **kwargs):
        super().__init__(weight,height)
        self.trunk_size = kwargs.get("trunk_size")   # can be skipped
        
    def make_sound(self):
        print("hreeee")

mangu = elephant(10000, 10)
mangu.height
print(vars(mangu))
```

```python
from abc import  ABC, abstractmethod

class novel(ABC):
    @abstractmethod
    def get_book_size(self):
        pass

class fictional_novels(novel):
    def __init__(self, size):
        self.size = size

    def get_book_size(self):
        return self.size

star_wars=fictional_novels(10000)
star_wars.get_book_size()
```
