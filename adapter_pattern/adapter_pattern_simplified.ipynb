# this pattern is used where the existing classes and method are not subject to modification
## also, this is used where we have an interface and it was doing something, but now the interface or class cannot be modified
## NOTE that the following examples uses three classes in total, but we can also stick with 2 class (1 new class + 1 adapter class + 1 old interface)

```python
class telephone():
    def __init__(self):
        pass

    def wired_call(self, telephone_number):
        print(f'calling the telephone_number {telephone_number}')

tata_wire_phone = telephone()
tata_wire_phone.wired_call("2544876")
```

```python
def call_a_person(medium, number):
    medium.wired_call(number)

call_a_person(tata_wire_phone, "2544876")
```

## after a decade, a new class has come and call_a_person will fail because the new class has new interfaces

```python
class mobile_phone():
    def __init__(self):
        pass

    def wireless_call(self, mobile_phone_number):
        print(f'calling the mobile_phone_number {mobile_phone_number}')
    
```

```python
class medium_adapter():
    def __init__(self, new_class):
        self.new_class = new_class

    def wired_call(self, number):
        self.new_class.wireless_call(number)

adapter = medium_adapter(mobile_phone())
call_a_person(adapter, "8222660874")
```

## both telephone and mobile_phone classes are not changed, only the calling is changed with the help of new class.
## Even the calling function is not changed, all are handled by adapter
### Assume that the telephone class, mobile_phone class and the calling method are provided by a library and you are not allowed to change neither of these. Hence the trick is to create a new Adapter class (by user himself) that utilizes the new class, and just calls a different underlying function(in the adapter class), and it looks like we are still using the call_a_person function.
