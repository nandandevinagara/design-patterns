## decorator pattern is mainly used to add/override behaviour of existing object

### note that the wrapper is called before the actual function, then actual function is also called
### Wrapper is called everytime instead of the actual function hence wrapper takes the same arguments as the function

```python
def decorator(func):
    print("in decorator")
    def wrapper():
        print('decorator internal function')
        func()

    return wrapper

@decorator                       # means you are calling the function within the decorator
def normal_func():
    print("normal function")

normal_func()
print("\n")
normal_func()
print("\n")
normal_func()
```

```python
def check_admin(func):

    def wrapper(user, path):
        if user == "admin":
            func()
        else:
            print("permission denied")

    return wrapper

@check_admin
def delete_file(user, path):
    print("delete the file")

delete_file('nandan', "abc/text/file1.txt")
```

```python
def add_enhancer(func):

    def decorator_internal_func( a, b):
        c = func(a, b)
        return c + 100
        
    return decorator_internal_func

@add_enhancer
def add(a, b):
    return a + b


add(2, 3)
```

```python
import time

def timer(func):
    def wrapper():
        t1 = time.time()
        func()
        print(f"time take to execute the function is {time.time() - t1}")
    return wrapper

@timer
def func():
    print("executing func")

func()
```

```python
def check_for_spaces(func):
    
    def wrapper(string):
        if " " in string:
            print("Invalid string, provide valid string")
        else:
            return func(string) # return is required or else the value is returned to wrapper, but wrapper is actually being called, so Print prints none
    return wrapper

@check_for_spaces
def to_uppercase(string):
    return string.upper()

print(to_uppercase("nandan"))
print(to_uppercase("nandan nayak"))  # with if satisfied, the wrapper doesnt return anything, hence None is printed
```
