```python
# "If it walks like a duck and quacks like a duck, treat it like a duck." 
```

```python
class oracle_db():
    def get_db_size(self):
        return 1000

class ms_db():
    def get_db_size(self):
        return 250

def get_db_size(db_type): # BASICALLY HERE ANY TYPE CAN BE SENT AS LONG THE FUNCTION HANDLES THE TYPE; ITS NOT POSSIBLE IN OTHER LANGUAGES DUE TO STRICT TYPES
    return db_type.get_db_size()

get_db_size(ms_db())
```

```python
# in simple words, for a function argument etc, in python, you need not send the exact type, you can send any types as long the type is properly processed 
# by the function
```
