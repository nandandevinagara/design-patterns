```python
class process():
    def __init__(self, name):
        self.name = name

    def __enter__(self):
        print("initializing the environment")
        return self    # here it has to return self or else the vlc_player object shall be assigned value as None

    def __exit__(self, exc_type, exc_value, traceback):  # these arguments are must
        print("destroying the environment")

with process("vlc player") as vlc_player:
    print(id(vlc_player))
    print("executing the process")        

# the object is not destroyed
print(id(vlc_player))

# THE INTERNAL MECHANISM OF WHATS GOING ON WITH CONTEXT MANAGERS
# THE FOLLOWING CODE IS IMPLICITLY CALLED

#vlc_player = process("vlc player")
#vlc_player.__enter__()

# your code runs

#vlc_player.__exit__(...)
```
