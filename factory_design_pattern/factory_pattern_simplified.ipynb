## just a class that has logic to create new classes belonging to a particular family
## it contains the logic and returns the instances of classes
## popular example is transformer where user can just enter the model name and the model is returned

### factory class helps in creating new object and has all the logic

```python
from abc import ABC, abstractmethod
class vlm(ABC):
    def __init__(self, version, size, quantization = "4bit"):
        self.version = version
        self.size = size
        self.quantization = quantization

    @abstractmethod
    def generate(self):
        pass

class qwen_model(vlm):
    def generate(self):
        print("generating the content from qwen model")

class deepseek_model(vlm):
    def generate(self):
        print("generating the content from deepseek model")

class vlm_factory():

    def create_vlm_model(self, family_name, version, size, quantization):
        if family_name == "qwen":
            print(f"returning the qwen model of {version} version and of {size} size")
            return qwen_model(version, size)
            
        elif family_name == "deepseek":
            print(f"returning the deepseek model of {version} version and of {size} size")
            return deepseek_model(version, size)


factory = vlm_factory()

qwen_2_5_model = factory.create_vlm_model("qwen", "qwen2.5", "8B", "16bit")    # qwen might have its own factory to return these qwen's sub classes
qwen_7_5_model = factory.create_vlm_model("qwen", "qwen7.5", "3B", "32bit")    # or it just returns a simple object that has found qwen's weights of this 
                                                                               # configuration and manages that files. there could be a method to find the location of weights
deepseek_omni_model = factory.create_vlm_model("deepseek", "omni", "81B", "4bit")

qwen_2_5_model.generate()
deepseek_omni_model.generate()
    
```
